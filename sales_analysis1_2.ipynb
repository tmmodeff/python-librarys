```python
import numpy as np
```

```python
months = np.array(["jan","feb","march","april","may","june","july","august","sept","oct","nov","dec"])
```

```python
sales =[ ]
```

```python
print("Enter a Sales in Dollar For Each Months :-")
for month in months :
    value = float (input (f"{months}:"))
    sales.append(value)
```

```python
sale = np.array(sales)
```

```python
print("Company Sales Analysis ")
print("Total Sales Of The Year :",np.sum(sales),"$")
print("Higest Sales Of The Year :",np.max(sales),"$")
print("Lowest Sales Of The Year :",np.min(sales),"$")
print("Average Monthly Sales Of The Year :",np.mean(sales),"$")
```

```python
best_month = months[np.argmax(sale)]
worst_month = months[np.argmin(sale)]
```

```python
print("Best Month (Higest sale of Year)",best_month)
print("Worst Month (Lowest sale of Year)",worst_month)
```

```python
above_average = months[sales > np.mean(sales)]
below_average = months[sales < np.mean(sales)]
```

```python
print("Above average sale of Year (In months )",above_average)
print("below average sale of Year (In months )",below_average)
```
