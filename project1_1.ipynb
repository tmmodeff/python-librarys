```python
import numpy as np
```

```python
taxi =np.genfromtxt('nyc_taxis.csv',delimiter = ',',skip_header=1)  # skip_header=1 : skips first row 
                                                                    # delimiter=',': Tells NumPy that data values are separated by commas
```

```python
speed = taxi[:,7]/(taxi[:,8]/3600)  # in 7th and 8th column there is data of distance and lenght data is in secons so we divided with 3600
```

```python
mean_speed = speed.mean()     #average speed of taxi 
print (mean_speed)    
```
